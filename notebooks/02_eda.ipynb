```python
import pandas as pd

data_path = "../data/processed/student_performance_clean.csv"

df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

display(df.head())
```

```python
import matplotlib.pyplot as plt

study_levels = sorted(df["studytime"].unique())

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        df[df["studytime"] == level]["G3"]
        for level in study_levels
    ],
    tick_labels=study_levels
)

plt.xlabel("Study Time Level")
plt.ylabel("Final Grade (G3)")
plt.title("Study Time vs Final Student Performance")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    df["absences"],
    df["G3"],
    alpha=0.7
)

plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")
plt.title("Absences vs Final Student Performance")

plt.show()
```

```python
import matplotlib.pyplot as plt

failure_levels = sorted(df["failures"].unique())

plt.figure(figsize=(8, 5))

plt.boxplot(
    [
        df[df["failures"] == level]["G3"]
        for level in failure_levels
    ],
    tick_labels=failure_levels
)

plt.xlabel("Number of Previous Failures")
plt.ylabel("Final Grade (G3)")
plt.title("Previous Failures vs Final Student Performance")

plt.show()
```

```python
import matplotlib.pyplot as plt

gender_groups = [
    df[df["sex"] == "F"]["G3"],
    df[df["sex"] == "M"]["G3"]
]

plt.figure(figsize=(8, 5))

plt.boxplot(
    gender_groups,
    tick_labels=["Female", "Male"]
)

plt.xlabel("Gender")
plt.ylabel("Final Grade (G3)")
plt.title("Gender-wise Final Student Performance")

plt.show()
```

```python
def support_category(score):
    if score <= 9:
        return "High Support"
    elif score <= 13:
        return "Medium Support"
    else:
        return "Low Support"

df["Support_Level"] = df["G3"].apply(support_category)

print("Support Level created successfully!")
print(df["Support_Level"].value_counts())
```

```python
import matplotlib.pyplot as plt

support_counts = df["Support_Level"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    support_counts.index,
    support_counts.values
)

plt.xlabel("Support Level")
plt.ylabel("Number of Students")
plt.title("Student Support Level Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(
    df["age"],
    df["G3"],
    alpha=0.7
)

plt.xlabel("Age")
plt.ylabel("Final Grade (G3)")
plt.title("Age vs Final Student Performance")

plt.show()
```
