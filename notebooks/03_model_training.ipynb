```python
# ============================================
# STEP 2: Prepare data for Machine Learning
# ============================================

from sklearn.model_selection import train_test_split

# Target variable
y = df["G3"]

# Remove target and previous grades to avoid data leakage
# G1 and G2 are closely related to G3, so for our main model
# we will predict G3 using the other student features.
X = df.drop(columns=["G3", "G1", "G2"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget (G3) statistics:")
print(y.describe())

print("\nFeature columns:")
print(X.columns.tolist())
```

```python
# ============================================
# STEP 3: Convert categorical data + Train/Test Split
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical and numerical columns
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()
numerical_cols = X.select_dtypes(exclude=["object"]).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nNumerical columns:")
print(numerical_cols)

# Preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
        ("numerical", "passthrough", numerical_cols)
    ]
)

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n--- TRAIN / TEST SPLIT ---")
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTarget training samples:", y_train.shape[0])
print("Target testing samples:", y_test.shape[0])
```

```python
# ============================================
# STEP 4: Train 5 ML Regression Models
# ============================================

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.neighbors import KNeighborsRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Preprocessing
preprocessor_dense = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_cols
        ),
        (
            "numerical",
            "passthrough",
            numerical_cols
        )
    ]
)

# 5 required models
models = {
    "Linear Regression": LinearRegression(),

    "Decision Tree": DecisionTreeRegressor(
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        random_state=42
    ),

    "KNN": KNeighborsRegressor(
        n_neighbors=5
    )
}

results = []

for name, model in models.items():

    pipeline = Pipeline([
        ("preprocessing", preprocessor_dense),
        ("model", model)
    ])

    # Train model
    pipeline.fit(X_train, y_train)

    # Prediction
    y_pred = pipeline.predict(X_test)

    # Metrics
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)

    results.append({
        "Model": name,
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "R2 Score": round(r2, 3)
    })

# Create results table
results_df = pd.DataFrame(results)

print("MODEL PERFORMANCE")
print("=" * 60)

display(
    results_df.sort_values(
        by="R2 Score",
        ascending=False
    ).reset_index(drop=True)
)
```

```python
# ============================================
# STEP 5: Select the Best Model
# ============================================

# Sort models by R2 score
results_df = results_df.sort_values(
    by="R2 Score",
    ascending=False
).reset_index(drop=True)

print("MODEL RANKING")
print("=" * 60)

display(results_df)

# Best model name
best_model_name = results_df.loc[0, "Model"]

print("\nBest Model:", best_model_name)
print("Best R2 Score:", results_df.loc[0, "R2 Score"])
print("Best MAE:", results_df.loc[0, "MAE"])
print("Best RMSE:", results_df.loc[0, "RMSE"])
```

```python
# ============================================
# STEP 6: Hyperparameter Tuning
# ============================================

from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline

# Random Forest pipeline
rf_pipeline = Pipeline([
    ("preprocessing", preprocessor_dense),
    ("model", RandomForestRegressor(random_state=42))
])

# Parameters to test
param_grid = {
    "model__n_estimators": [100, 200],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_split": [2, 5]
}

# Grid Search
grid_search = GridSearchCV(
    estimator=rf_pipeline,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_absolute_error",
    n_jobs=-1
)

print("Starting Hyperparameter Tuning...")
print("Please wait...")

grid_search.fit(X_train, y_train)

print("\nHyperparameter tuning completed!")

print("\nBest Parameters:")
print(grid_search.best_params_)

print("\nBest Cross-Validation Score:")
print(-grid_search.best_score_)

# Best tuned model
best_model = grid_search.best_estimator_

# Test prediction
tuned_predictions = best_model.predict(X_test)

# Evaluation
tuned_mae = mean_absolute_error(y_test, tuned_predictions)
tuned_rmse = np.sqrt(
    mean_squared_error(y_test, tuned_predictions)
)
tuned_r2 = r2_score(y_test, tuned_predictions)

print("\n--- TUNED MODEL TEST PERFORMANCE ---")
print("MAE :", round(tuned_mae, 3))
print("RMSE:", round(tuned_rmse, 3))
print("R2  :", round(tuned_r2, 3))
```

```python
# ============================================
# STEP 9: Create Student Support Categories
# ============================================

def support_category(score):
    if score <= 9:
        return "High Support"
    elif score <= 13:
        return "Medium Support"
    else:
        return "Low Support"


# Create classification target
df["Support_Level"] = df["G3"].apply(support_category)

print("Support Level Distribution")
print("=" * 50)

print(df["Support_Level"].value_counts())

print("\nPercentage Distribution")
print(
    (df["Support_Level"].value_counts(normalize=True) * 100)
    .round(2)
)
```

```python
# ============================================
# STEP 10: Prepare Classification Dataset
# ============================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Features
X_class = df.drop(
    columns=["G3", "G1", "G2", "Support_Level"]
)

# Classification target
y_class = df["Support_Level"]

# Identify columns
categorical_class_cols = X_class.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_class_cols = X_class.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Categorical features:")
print(categorical_class_cols)

print("\nNumerical features:")
print(numerical_class_cols)

# Preprocessor
class_preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_class_cols
        ),
        (
            "numerical",
            "passthrough",
            numerical_class_cols
        )
    ]
)

# Train/Test split
Xc_train, Xc_test, yc_train, yc_test = train_test_split(
    X_class,
    y_class,
    test_size=0.20,
    random_state=42,
    stratify=y_class
)

print("\n--- CLASSIFICATION SPLIT ---")
print("Training samples:", Xc_train.shape[0])
print("Testing samples:", Xc_test.shape[0])

print("\nTraining class distribution:")
print(yc_train.value_counts())

print("\nTesting class distribution:")
print(yc_test.value_counts())
```

```python
# ============================================
# STEP 11: Train 5 Classification Models
# ============================================

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# 5 classification models
classification_models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        random_state=42
    ),

    "KNN": KNeighborsClassifier(
        n_neighbors=5
    )
}

classification_results = []

for name, model in classification_models.items():

    pipeline = Pipeline([
        ("preprocessing", class_preprocessor),
        ("model", model)
    ])

    # Train
    pipeline.fit(Xc_train, yc_train)

    # Predict
    yc_pred = pipeline.predict(Xc_test)

    # Metrics
    accuracy = accuracy_score(yc_test, yc_pred)

    precision = precision_score(
        yc_test,
        yc_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        yc_test,
        yc_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        yc_test,
        yc_pred,
        average="weighted",
        zero_division=0
    )

    classification_results.append({
        "Model": name,
        "Accuracy": round(accuracy, 3),
        "Precision": round(precision, 3),
        "Recall": round(recall, 3),
        "F1 Score": round(f1, 3)
    })


# Results table
classification_results_df = pd.DataFrame(
    classification_results
)

classification_results_df = classification_results_df.sort_values(
    by="F1 Score",
    ascending=False
).reset_index(drop=True)

print("CLASSIFICATION MODEL PERFORMANCE")
print("=" * 70)

display(classification_results_df)
```

```python
# ============================================
# STEP 12: Confusion Matrix + Classification Report
# ============================================

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
import matplotlib.pyplot as plt

# Best model name
best_classification_name = classification_results_df.loc[0, "Model"]

print("Best Classification Model:")
print(best_classification_name)

# Get the corresponding model
best_classification_model = classification_models[
    best_classification_name
]

# Create pipeline
best_classification_pipeline = Pipeline([
    ("preprocessing", class_preprocessor),
    ("model", best_classification_model)
])

# Train best model
best_classification_pipeline.fit(
    Xc_train,
    yc_train
)

# Predictions
yc_best_pred = best_classification_pipeline.predict(
    Xc_test
)

# Classification report
print("\nCLASSIFICATION REPORT")
print("=" * 70)

print(
    classification_report(
        yc_test,
        yc_best_pred,
        zero_division=0
    )
)

# Confusion Matrix
cm = confusion_matrix(
    yc_test,
    yc_best_pred,
    labels=["High Support", "Medium Support", "Low Support"]
)

print("\nCONFUSION MATRIX")
print(cm)

# Plot
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=[
        "High Support",
        "Medium Support",
        "Low Support"
    ]
)

disp.plot()

plt.title(
    f"Confusion Matrix - {best_classification_name}"
)

plt.show()
```

```python
# ============================================
# STEP 13: Student Engagement Clustering
# ============================================

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Select engagement-related numerical features
cluster_features = [
    col for col in [
        "studytime",
        "failures",
        "absences",
        "Medu",
        "Fedu",
        "goout",
        "Dalc",
        "Walc",
        "health"
    ]
    if col in df.columns
]

print("Features used for clustering:")
print(cluster_features)

# Create clustering dataset
cluster_data = df[cluster_features].copy()

# Scale features
scaler = StandardScaler()

cluster_scaled = scaler.fit_transform(
    cluster_data
)

# Try different number of clusters
silhouette_scores = {}

for k in range(2, 7):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(cluster_scaled)

    score = silhouette_score(
        cluster_scaled,
        labels
    )

    silhouette_scores[k] = score

print("\nSilhouette Scores")
print("=" * 50)

for k, score in silhouette_scores.items():
    print(
        f"Clusters = {k}  |  "
        f"Silhouette Score = {score:.3f}"
    )

# Select best K
best_k = max(
    silhouette_scores,
    key=silhouette_scores.get
)

print("\nBest number of clusters:", best_k)

# Final K-Means model
kmeans_final = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df["Student_Cluster"] = kmeans_final.fit_predict(
    cluster_scaled
)

print("\nCluster Distribution")
print("=" * 50)

print(
    df["Student_Cluster"].value_counts().sort_index()
)
```

```python
# ============================================
# STEP 14: Analyze Student Clusters
# ============================================

import matplotlib.pyplot as plt

# Cluster-wise average values
cluster_summary = df.groupby("Student_Cluster")[
    cluster_features
].mean().round(2)

print("CLUSTER PROFILE SUMMARY")
print("=" * 60)

display(cluster_summary)


# --------------------------------------------
# Cluster size
# --------------------------------------------

print("\nSTUDENTS IN EACH CLUSTER")
print("=" * 60)

cluster_counts = df["Student_Cluster"].value_counts().sort_index()

display(cluster_counts)


# --------------------------------------------
# Visualization: Cluster sizes
# --------------------------------------------

plt.figure(figsize=(7, 5))

plt.bar(
    cluster_counts.index.astype(str),
    cluster_counts.values
)

plt.xlabel("Student Cluster")
plt.ylabel("Number of Students")
plt.title("Student Engagement Cluster Distribution")

plt.show()
```

```python
# ============================================
# STEP 15: Personalized Learning Recommendations
# ============================================

def generate_recommendations(student_data):
    """
    Generate simple personalized learning recommendations
    based on student academic and engagement indicators.
    """

    recommendations = []

    # Attendance / absence
    if "absences" in student_data:
        if student_data["absences"] >= 10:
            recommendations.append(
                "Improve attendance and maintain a regular study schedule."
            )
        elif student_data["absences"] >= 5:
            recommendations.append(
                "Try to reduce absences and attend classes more consistently."
            )

    # Previous failures
    if "failures" in student_data:
        if student_data["failures"] >= 2:
            recommendations.append(
                "Review previously difficult subjects and practice basic concepts."
            )
        elif student_data["failures"] == 1:
            recommendations.append(
                "Spend additional time revising topics where you previously struggled."
            )

    # Study time
    if "studytime" in student_data:
        if student_data["studytime"] <= 2:
            recommendations.append(
                "Increase weekly study time using short, consistent study sessions."
            )
        else:
            recommendations.append(
                "Continue the current study routine and focus on weak topics."
            )

    # Social activity
    if "goout" in student_data:
        if student_data["goout"] >= 4:
            recommendations.append(
                "Balance social activities with a structured academic schedule."
            )

    # Alcohol-related indicators
    if "Dalc" in student_data and student_data["Dalc"] >= 3:
        recommendations.append(
            "Maintain healthy routines and minimize factors that may affect study consistency."
        )

    if "Walc" in student_data and student_data["Walc"] >= 3:
        recommendations.append(
            "Maintain healthy routines and protect regular study time."
        )

    # Default recommendation
    if len(recommendations) == 0:
        recommendations.append(
            "Maintain your current learning routine and regularly review your progress."
        )

    return recommendations


# --------------------------------------------
# Test recommendation system
# --------------------------------------------

sample_student = df.iloc[0].to_dict()

recommendations = generate_recommendations(
    sample_student
)

print("PERSONALIZED LEARNING RECOMMENDATIONS")
print("=" * 60)

for i, recommendation in enumerate(
    recommendations,
    start=1
):
    print(f"{i}. {recommendation}")
```

```python
import pandas as pd
import numpy as np

def explain_student(model, student_row, reference_data, top_n=5):
    original_prediction = model.predict(student_row)[0]

    changes = []

    for feature in student_row.columns:
        modified_row = student_row.copy()

        if pd.api.types.is_numeric_dtype(reference_data[feature]):
            replacement = reference_data[feature].median()
        else:
            replacement = reference_data[feature].mode()[0]

        modified_row[feature] = replacement

        new_prediction = model.predict(modified_row)[0]

        changes.append({
            "Feature": feature,
            "Original Value": student_row[feature].iloc[0],
            "Reference Value": replacement,
            "Prediction Change": new_prediction - original_prediction,
            "Absolute Impact": abs(new_prediction - original_prediction)
        })

    explanation_df = pd.DataFrame(changes)
    explanation_df = explanation_df.sort_values(
        "Absolute Impact",
        ascending=False
    ).head(top_n)

    return original_prediction, explanation_df
```
